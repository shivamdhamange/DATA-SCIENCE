## 0. Load and inspect the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("D:\data analyst\pandas\DATA\DataFrame\movies.csv")

print("Rows and columns:", df.shape)
print("Column names:", df.columns.tolist())
display(df.head())
```

## 1. Line Chart

Question: Plot the IMDb ratings of the first 10 movies with non-missing ratings. Add a title, axis labels, markers, rotated movie names, and a grid.

Answer Code

```python
sample = df[["title_x", "imdb_rating"]].dropna().head(10)

plt.figure(figsize=(10, 5))
plt.plot(sample["title_x"], sample["imdb_rating"],
         color="blue", marker="o", linestyle="-")
plt.title("IMDb Ratings of First 10 Movies")
plt.xlabel("Movie Name")
plt.ylabel("IMDb Rating")
plt.xticks(rotation=45, ha="right")
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Interpretation note:This chart is for practicing line-chart syntax. Movie titles are categories, so this is not a time-trend chart.

## 2. Vertical Bar Chart

Question: Count the first-listed genre for each movie record and show the top 10 genres in a vertical bar chart. Add labels and a horizontal grid.

Answer Code

```python
genre_counts = (
    df["genres"].dropna()
      .str.split("|").str[0]
      .value_counts()
      .head(10)
)

plt.figure(figsize=(10, 5))
plt.bar(genre_counts.index, genre_counts.values,
        color="skyblue", edgecolor="black")
plt.title("Top 10 First-Listed Genres")
plt.xlabel("First-Listed Genre")
plt.ylabel("Number of Movie Records")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

**Important:** This simplified exercise counts only the first-listed genre in each row, not every genre attached to a movie.

## 3. Horizontal Bar Chart

Question: Display the same top 10 first-listed genres using a horizontal bar chart.

Answer Code

```python
plt.figure(figsize=(9, 6))
plt.barh(genre_counts.index, genre_counts.values,
         color="orange", edgecolor="black")
plt.title("Top 10 First-Listed Genres")
plt.xlabel("Number of Movie Records")
plt.ylabel("First-Listed Genre")
plt.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## 4. Scatter Plot

Question: Explore the relationship between IMDb votes and IMDb rating. Plot `imdb_votes` on the X-axis and `imdb_rating` on the Y-axis. Use transparency and labels.

Answer Code

```python
scatter_data = df[["imdb_votes", "imdb_rating"]].dropna()

plt.figure(figsize=(8, 5))
plt.scatter(scatter_data["imdb_votes"], scatter_data["imdb_rating"],
            color="green", s=35, alpha=0.5)
plt.title("IMDb Votes vs IMDb Rating")
plt.xlabel("IMDb Votes")
plt.ylabel("IMDb Rating")
plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

**Interpretation note:** A scatter plot helps explore association; it does not prove that votes cause a rating to rise or fall.

## 5. Histogram

Question: Show the distribution of IMDb ratings using 10 bins.

Answer Code

```python
ratings = df["imdb_rating"].dropna()

plt.figure(figsize=(8, 5))
plt.hist(ratings, bins=10, color="mediumpurple", edgecolor="black")
plt.title("Distribution of IMDb Ratings")
plt.xlabel("IMDb Rating")
plt.ylabel("Number of Movie Records")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

## 6. Pie Chart

Question: Show the top 5 first-listed genres as slices in a pie chart, including percentages and a legend.



```python
top5_genres = genre_counts.head(5)

plt.figure(figsize=(8, 6))
plt.pie(top5_genres.values,
        labels=top5_genres.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Top 5 First-Listed Genres")
plt.legend(title="Genres", loc="best")
plt.tight_layout()
plt.show()
```

**Interpretation note:** Percentages are calculated among these selected top 5 genre counts, not across all genre categories.

## 7. Box Plot

Question: Create a box plot for IMDb ratings. Add a title and grid.

Answer Code

```python
ratings = df["imdb_rating"].dropna()

plt.figure(figsize=(8, 5))
plt.boxplot(ratings)
plt.title("Box Plot of IMDb Ratings")
plt.ylabel("IMDb Rating")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

Key terms: The line inside the box is the median. The box covers Q1 to Q3. Points beyond the whiskers may be flagged as outliers according to the boxplot rule.

## 8. Area Chart

Question: Calculate the average IMDb rating by release year and show the trend as a line with the area underneath filled.



```python
year_rating = (
    df[["year_of_release", "imdb_rating"]]
    .dropna()
    .groupby("year_of_release", as_index=False)["imdb_rating"]
    .mean()
    .sort_values("year_of_release")
)

plt.figure(figsize=(10, 5))
plt.plot(year_rating["year_of_release"], year_rating["imdb_rating"],
         color="darkgreen", marker="o", linewidth=1.5)
plt.fill_between(year_rating["year_of_release"], year_rating["imdb_rating"],
                 alpha=0.2, color="green")
plt.title("Average IMDb Rating by Release Year")
plt.xlabel("Year of Release")
plt.ylabel("Average IMDb Rating")
plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 9. Subplots: Two Charts in One Figure

Question: Create two charts side by side: a histogram of IMDb ratings and a bar chart of the top 5 first-listed genres.


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df["imdb_rating"].dropna(), bins=10,
             color="skyblue", edgecolor="black")
axes[0].set_title("IMDb Rating Distribution")
axes[0].set_xlabel("IMDb Rating")
axes[0].set_ylabel("Number of Movie Records")

axes[1].bar(top5_genres.index, top5_genres.values,
            color="orange", edgecolor="black")
axes[1].set_title("Top 5 First-Listed Genres")
axes[1].set_xlabel("Genre")
axes[1].set_ylabel("Number of Movie Records")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```
