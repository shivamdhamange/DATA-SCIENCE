```python
import numpy as np
```

# Indexing

# 1 . Positive indexing

array   -------     ["Apple","Banana","Grapes","Orange","Strawberry"]
#
positive indexing  ---0--------1--------2--------3---------4--------

```python
fruits = np.array(["Apple","Banana","Grapes","Orange","Strawberry"])
print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[3])
print(fruits[4])
```

# 2 . negative indexing

list   -------     ["Apple","Banana","Grapes","Orange","Strawberry"]
#
negative indexing  --(-5)----(-4)----(-3)-----(-2)------(-1)--------

```python
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
print(fruits[-4])
print(fruits[-5])
```

# Slicing

arr[start:end:step]

```python
print(fruits[1:4])
print(fruits[2:4])
print(fruits[-4:-1])
```

2D ARRAY

```python
marks = np.array([
    [45,69],
    [88,23]
])

print(marks[0])
print(marks[1])
```

ADDITION

```python
a = np.array([10,20,30])
b = np.array([40,50,60])
print(a+b)
```

# Arithmatic operation

COMPARISON OPERATOR

```python
marks = np.array([35,55,70,40,90])
print(marks>50)
print(marks<50)
print(marks == 70)
```

BOOLEAN MASKING

```python
marks = np.array([35,55,70,40,90])
mask = marks>50
print(mask)
print(marks[mask])
```

```python
# example for students marks
import numpy as np

names = np.array(["Shivam","Falguni","Danish","Athar","Faizan"])
marks = np.array([89,55,69,88,14])

passed = marks >= 40

print(f"Passed Students : {names[passed]}")
print(f"Pass Marks : {marks[passed]}")
print(f"Failed Students : {names[~passed]}")
```
