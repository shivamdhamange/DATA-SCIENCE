# Broadcasting
job numpy different shapes ke arrays par
arthematic operations perform karta hai to compatible #
shapes ko automatically adjust karke calculation karta hai

```python
import numpy as np
arr = np.array([10,20,30,40])
result = arr + 5
result
```

Broadcasting with 2-d array

```python
arr = np.array([
    [10,20,30],
    [40,50,60]]
)
result = arr + 5
result
```

real world problem

```python
# sales target
# company ne har product ke liye 1000 rs additional target set kiya
sales = np.array([[5000,6000,7000],
                  [8000,9000,10000]])
target = sales + 1000
target
```

broadcasting with two array

```python
prices = np.array([100,200,300])
discount = np.array([50,70,10])
result = prices - discount
result
```
