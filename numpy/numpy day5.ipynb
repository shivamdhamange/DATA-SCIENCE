Reshape  kya hota hai
array ke elements ko change kiye bina uska structure

```python
import numpy as np
arr = np.array([1,2,3,4,5,6])
newArray = arr.reshape(2,3)
newArray

# one dimension ko 2 rows and 3 columns me reshape liya
```

```python
arr = np.arange(1,13)
arr.reshape(3,4)
arr.reshape(4,3)
arr.reshape(2,2,3)

# reshape karte time total element same hona chahiye
# 12 element -- reshape (3,4) valid
# 12 element -- reshape (2,6) valid
# 12 element -- reshape (3,5) invalid
```

```python
arr = np.arange(1,13)
arr.reshape(3,-1)

# it will automatically detect the remaing dimension if we give -1
```
