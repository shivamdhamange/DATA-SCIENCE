# Broadcasting and Transpose

```python
import numpy as np
# transpose in numpy
# rows ko column aur colums ko rows me convert krta h
arr =np.array([[1,2,3],
              [4,5,6]])
result=arr.T         #(T is for transpose)
print(result)
print(arr.shape)
print(result.shape)
```

```python
#Q2)
marks= np.array([
    [80,85,90],
    [70,75,80],
    [90,95,85]

])
print("original:")
print(marks)
print("transpose:")
print(marks.T)
```

```python
"""Problem: Company ke paas 3 employees 
aur 3 months ki sales hain.
Humein har employee ki monthly sales mein
2,000 add karna hai."""

import numpy as np

sales = np.array([
    [10000, 12000, 15000],
    [20000, 22000, 25000],
    [15000, 18000, 20000]
])

# Broadcasting
updated_sales = sales + 2000

# Shape
print("Shape:", updated_sales.shape)

# Transpose
print("Transpose:")
print(updated_sales.T)

# Flatten
print("Flatten:")
print(updated_sales.flatten())
```
