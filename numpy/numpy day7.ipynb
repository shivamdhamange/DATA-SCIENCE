```python
import numpy as np
```

Concatination  -  combine two or more array into single array

syntax
np.concatinate((array1,array2),axis=0)
array1,array2  array to combine
axis =0   combine along rows
axis =1   combine along columns

```python
arr1 = np.array([100,200,300])
arr2 = np.array([400,500,600])

arr = np.concatenate((arr1,arr2))  # default
#arr = np.concatenate((arr1,arr2),axis=0)

print(arr)
```

Concate for two d array

```python
arr1 = np.array([[1,2],[3,4]])
arr2 = np.array([[7,8],[9,3]])

res = np.concatenate((arr1,arr2))
#res = np.concatenate((arr1,arr2),axis=0)  #combine along rows
#res = np.concatenate((arr1,arr2),axis=1)  #combine along columns
print(res)
```

```python
result = np.concatenate((arr1,arr2),axis=1)  #combine along columns
print(result)
```

# Stack

stack combine array along a new axis
stck creates new dimnsion while combininig arrays

syntax
np.stack((arr1,arr2),axis=0)

```python
student1 = np.array([80,90,70])
student2 = np.array([70,30,20])
result = np.stack((student1,student2),axis=0)

print(result)
print("Shape : ",result.shape)
```

```python
# stack with axis = 1
student1 = np.array([80,90,70])
student2 = np.array([70,30,20])
result = np.stack((student1,student2),axis=1)

print(result)
print("Shape : ",result.shape)
```

hstack and vstack
ye stack ke related functions hai


hstack == horizontal stacking


```python
a = np.array([10,20,30])
b = np.array([40,50,60])

print(np.hstack((a,b)))
```
