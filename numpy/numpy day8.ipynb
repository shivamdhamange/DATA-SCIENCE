# SPLIT



np.split(arr,3)  --- 3 equal parts
np.split(arr,[2,4])  --- split at index 2 and 4

```python
# divide in equal parts

import numpy as np
student = np.array([10,20,30,40,50,60])
result = np.split(student,3)
print(result)
```

```python
# split using indices

student = np.array([10,20,30,40,50,60])
res = np.split(student,[1,4])
print(res)
```

Split in 2d arrays

```python
arr = np.array([
    [10,20,30,89],
    [40,50,60,34]
])
print(np.hsplit(arr,4))
```

```python
arr = np.array([
    [10,20,30,89],
    [40,50,60,34]
])
print(np.vsplit(arr,2))
```

programming mein iski ko copy kehte hein

copy

```python
arr = np.array([10,20,30])
new_arr = arr.copy()
new_arr[0] = 100
print("Original: ",arr)
print("Copy: ",new_arr)
```

view

```python
arr = np.array([10,20,30])
new_arr = arr.view()
new_arr[0] = 100
print("Original: ",arr)
print("Copy: ",new_arr)
```
