JOIN TWO DIFFERENT TABLES USING JOIN

```python
movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy"]
}, index=["Dangal", "Jawan", "3 Idiots"])

print(movies)
```

```python
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.4]
}, index=["Dangal", "Jawan", "3 Idiots"])

print(ratings)
```

```python
#left join
```

```python
print(movies.join(ratings,how='left'))
```

```python
#right join
print(movies.join(ratings,how='right'))
```

```python
# inner join
print(movies.join(ratings,how='inner'))
```

how = 
left - pahale table ke sare record
right - right table index
inner - sirf matching index
outer - dono table ke sabhi index

```python
# Table 1: Movie aur Genre
mov = csv_movies[["Series_Title", "Genre"]].copy()

# Table 2: Movie aur Rating
rat = csv_movies[["Series_Title", "IMDB_Rating"]].copy()

# Movie name ko index banayein
mov = mov.set_index("Series_Title")
rat = rat.set_index("Series_Title")

# Dono tables join karein
result = mov.join(rat)

print(result)

```

```python
import pandas as pd

# Table 1: Movie details
movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy", "Thriller"],
    "Year": [2016, 2023, 2009, 2022]
}, index=["Dangal", "Jawan", "3 Idiots", "Drishyam"])

# Table 2: Movie ratings
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.4, 8.2],
    "Votes": [2000, 1500, 1800, 1200]
}, index=["Dangal", "Jawan", "PK", "Drishyam"])

print("MOVIES TABLE")
print(movies)

print("\nRATINGS TABLE")
print(ratings)
```

```python
# left join
r2 = movies.join(ratings,how = "left")
r2
```

```python
# right join
r3 = movies.join(ratings,how = "right")
r3
```

```python
# inner join
r4 = movies.join(ratings,how = "inner")
r4
```

```python
# outer join
r5 = movies.join(ratings,how = "outer")
r5
```
