Concat()
Recipe website ki recipe list do alag parts mein mili hai. Manager ko dono lists ko ek list mein combine karna hai
Do DataFrames hain. Kya humein 
rows jodni hain ya columns?
Dono DataFrames mein same columns hone chahiye ya different?
Agar pehli list mein 5 rows aur doosri mein 5 rows hain, toh combine karne par kitni rows hongi?

```python
first_list = df[["recipe_name", "rating"]].iloc[:5]
second_list = df[["recipe_name", "rating"]].iloc[5:10]

print(first_list)
print(second_list)

combined = pd.concat(
    [first_list, second_list],
    ignore_index=True
)

print(combined)
print(combined.shape)
```

Merge

```python
recipe_info = (
    df[["recipe_name","servings"]].drop_duplicates("recipe_name")
)

recipe_ratings = (
    df[["recipe_name","rating"]]
    .drop_duplicates("recipe_name")
)
result = pd.merge(recipe_info,recipe_ratings,on="recipe_name",how="inner")
print(result.head())
```

Pivot Table

```python
df["category"] = (
    df["cuisine_path"]
    .fillna("")
    .str.strip("/")
    .str.split("/")
    .str[0]
)

summary = pd.pivot_table(
    df[df["category"] != ""],
    index = "category",
    values="rating",
    aggfunc="mean"
)

print(summary.sort_values("rating",ascending=False))
```

```python
count_summary = pd.pivot_table(
    df[df["category"]!= ""],
    index = "category",
    values="recipe_name",
    aggfunc="count"
)

print(count_summary)
```

```python
#desert ki average rating nikalani hai toh index values aggfunc

desert_average = pd.pivot_table(
    df[df["category"]== "Desserts"],
    index = "category",
    values="rating",
    aggfunc="mean"
)

print(desert_average)
```
