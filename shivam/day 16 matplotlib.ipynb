# LINE CHART

# EXAMPLE 1

```python
movies = ["Movie A","Movie B","Movie C","Movie D","Movie E"]  # x axis
ratings = [6.5,8.0,7.2,9.0,7.8]    # y axis
 
plt.plot(movies,ratings)  # line chart banata hai      plt.plot(movies,ratings)

plt.show()  # chart display karta hai


#movies -- X axis par movie names
# ratings -- Y axis par ratings hai
```

```python
plt.figure(figsize=(9,4))  # chart ka size set karta hai

plt.plot(
    movies,
    ratings,
    marker="o",
    label = "IMDB Rating"
)

plt.title("Movie Rating Comparison")
plt.xlabel("Movie Name")
plt.ylabel("IMDB Rating")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

plt.figure(figsize=(9, 4))	Chart ka size set karta hai
plt.title()	Chart ka title
plt.xlabel()	X-axis ka naam
plt.ylabel()	Y-axis ka naam
marker="o"	    Har data point par circle
plt.legend()	Data series ka label
plt.grid(True)	Reference grid lines
plt.tight_layout()	Labels ke liye layout adjust karta hai

plt.ylabel("IMDb Rating") hata dein, toh kya chart phir bhi banega? Haan. Chart banega, lekin Y-axis ka label nahi dikhega.

# MOVIES.CSV FILE USE

Question: movies.csv dataset mein se title_x aur imdb_rating columns select karke pehli 10 movies ki IMDb ratings ka Line Chart banao. Chart ka title, X-axis label, Y-axis label aur rotated movie names add karo.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\ITP\DATA SCIENCE\data\movies (1).csv")

sample = df[["title_x", "imdb_rating"]].dropna().head(10)

plt.figure(figsize=(10, 5))

plt.plot(
    sample["title_x"],
    sample["imdb_rating"],
    color="blue",
    marker="o",
    linestyle="-"
)

plt.title("IMDb Ratings of First 10 Movies")
plt.xlabel("Movie Name")
plt.ylabel("IMDb Rating")

plt.xticks(rotation=45, ha="right")
plt.grid(True)
plt.tight_layout()
plt.show()
```

# BAR CHART

# EXAMPLE 1

Movie Genres ka Bar Chart
Task: movies.csv dataset ka use karke pehli listed genre ke according top 10 movie genres count karo aur vertical Bar Chart banao. Chart mein title, axis labels aur grid add karo.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\ITP\DATA SCIENCE\data\movies (1).csv")

genre_counts = (
    df["genres"]
    .dropna()
    .str.split("|")
    .str[0]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 5))

plt.bar(
    genre_counts.index,
    genre_counts.values,
    color="skyblue",
    edgecolor="black"
)

plt.title("Top 10 Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Movie Records")

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

# HISTOGRAM 

# Example 1

Histogram — Question & Answer
Question: movies.csv dataset ke imdb_rating column ka Histogram banao. Chart mein title, X-axis label, Y-axis label aur grid add karo.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\ITP\DATA SCIENCE\data\movies (1).csv")

ratings = df["imdb_rating"].dropna()

plt.figure(figsize=(8, 5))

plt.hist(
    ratings,
    bins=10,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of IMDb Ratings")
plt.xlabel("IMDb Rating")
plt.ylabel("Number of Movies")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

dropna() — missing ratings remove karta hai.
bins=10 — ratings ki range ko 10 intervals mein divide karta hai.
plt.hist() — Histogram banata hai.
edgecolor="black" — bars ke borders black karta hai.
plt.ylabel() — har range mein kitne movie records hain, woh dikhata hai.

# PIE CHART

# Example 1

Movie Genres ka Pie Chart
Question: movies.csv se genres count karo aur top 5 first-listed genres ka Pie Chart banao. Har slice par percentage aur legend bhi dikhana hai.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\ITP\DATA SCIENCE\data\movies (1).csv")

genre_counts = (
    df["genres"]
    .dropna()
    .str.split("|")
    .str[0]
    .value_counts()
    .head(5)
)

plt.figure(figsize=(8, 6))

plt.pie(
    genre_counts.values,
    labels=genre_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Movie Genres")
plt.legend(title="Genres")
plt.tight_layout()
plt.show()
```

plt.pie() Pie Chart banata hai
labels= Har slice ka naam
autopct="%1.1f%%" Percentage ek decimal place tak dikhata hai
startangle=90 Chart ka starting angle set karta hai
plt.legend() Legend dikhata hai
head(5) Top 5 genres select karta hai

# Example 2

Pie Chart ke colors change karo
Task: Har genre ko alag color do.

```python
genre_counts = (
    df["genres"]
    .dropna()
    .str.split("|")
    .str[0]
    .value_counts()
    .head(5)
)


plt.figure(figsize=(8, 6))

plt.pie(
    genre_counts.values,
    labels=genre_counts.index,
    colors=["red", "orange", "skyblue", "green", "purple"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Movie Genre Distribution")
plt.tight_layout()
plt.show()
```

BOX PLOT

# Example 1

Box Plot — Question & Answer
Question 1: Movies ki IMDb ratings ka Box Plot banao
movies.csv dataset ke imdb_rating column ka Box Plot banao. Chart mein title, axis label aur grid add karo.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"D:\ITP\DATA SCIENCE\data\movies (1).csv")

ratings = df["imdb_rating"].dropna()

plt.figure(figsize=(8, 5))

plt.boxplot(ratings)

plt.title("Distribution of Movie IMDb Ratings")
plt.ylabel("IMDb Rating")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

# Example 2

Horizontal Box Plot banao
Task: Usi ratings ka horizontal Box Plot banao.

```python
ratings = df["imdb_rating"].dropna()

plt.figure(figsize=(8, 4))

plt.boxplot(ratings, vert=False)

plt.title("Horizontal Box Plot of IMDb Ratings")
plt.xlabel("IMDb Rating")
plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
