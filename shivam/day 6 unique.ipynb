**DATA CLEANING**

Isnull()


NONe --- data avaible nahi hota hai/not a number

```python
import pandas as pd

data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105"],
    "Age": [45, 62, None, 55, 41],
    "City": ["Hyderabad", "Warangal", "Hyderabad", None, "Hyderabad"],
    "Operation_Duration_Min": [90, 240, 75, None, 180]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull())

# true shows value missing
# falsee shows values that are available
```

```python
print(df['Age'].isnull())
```
