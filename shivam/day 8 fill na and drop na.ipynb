```python
import pandas as pd

data = {
    "Patient_ID": [101,102,103,104,105,106,107,108,109,110],
    "Age": [25,45,35,60,50,28,65,40,55,30],
    "Gender": ["Male","Female","Male","Female","Male",
               "Female","Male","Female","Male","Female"],
    "Department": ["Cardiology","Neurology","Cardiology","Orthopedic",
                   "Neurology","Cardiology","Orthopedic","Neurology",
                   "Cardiology","Orthopedic"],
    "City": ["Nagpur","Pune","Nagpur","Mumbai","Pune",
             "Nagpur","Mumbai","Pune","Nagpur","Mumbai"],
    "Treatment_Cost": [50000,35000,60000,45000,30000,
                       55000,70000,40000,65000,50000],
    "Hospital_Stay": [5,3,6,4,2,5,7,3,6,4]
}

df = pd.DataFrame(data)

df
```

FILL NA AND DROP NA

```python
import pandas as pd

data = {
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "Age": [25, 45, None, 60, 50, 28, 65, None, 55, 30],

    "Gender": [
        "Male", "Female", "Male", "Female", None,
        "Female", "Male", "Female", "Male", "Female"
    ],

    "Department": [
        "Cardiology", "Neurology", "Cardiology", "Orthopedic",
        "Neurology", "Cardiology", "Orthopedic", "Neurology",
        "Cardiology", "Orthopedic"
    ],

    "City": [
        "Nagpur", "Pune", "Nagpur", None,
        "Pune", "Nagpur", "Mumbai", "Pune",
        "Nagpur", "Mumbai"
    ],

    "Treatment_Cost": [
        50000, 35000, None, 45000, 30000,
        55000, 70000, 40000, None, 50000
    ],

    "Hospital_Stay": [
        5, 3, 6, None, 2,
        5, 7, 3, 6, 4
    ]
}

df = pd.DataFrame(data)

df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

DROPNA()  remove missing row

```python
df.dropna()   # missing record hatane ke liye
# it check fo rna in hole data
```

```python
df.dropna(subset=["Age"])   # using subset it check only in single column for na
```

FILL NA


```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```
