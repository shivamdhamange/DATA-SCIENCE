LOC FUNCTION


PANDAS NOTES — LOC

1️⃣ loc — Select Data Based on Labels or Conditions

Concept

loc is used to select rows and columns based on row/column labels or conditions.

 Basic Syntax:

df.loc[row, column]


2️⃣ Select a Specific Row

If we want the row with index 0:

df.loc[0]

This displays the complete row at index 0.


3️⃣ Select a Specific Row + Specific Column

To get the Patient_ID of the patient at index 0:

df.loc[0, "Patient_ID"]

Here:

0 → Row Index
"Patient_ID" → Column Name


4️⃣ Select Multiple Columns Using loc

To display Patient_ID and Surgery_Type for all patients:

df.loc[:, ["Patient_ID", "Surgery_Type"]]

Meaning of :

: = All Rows

So:

df.loc[:, ["Patient_ID", "Surgery_Type"]]

means:

All Rows + Selected Columns


5️⃣  loc with Condition

This is an important Data Analyst use-case.

To find patients from Hyderabad:

df.loc[df["City"] == "Hyderabad"]

Here:

df["City"] == "Hyderabad"

= Condition


6️⃣ Condition + Selected Columns

To find Hyderabad patients and display only selected information:

df.loc[
    df["City"] == "Hyderabad",
    ["Patient_ID", "Surgery_Type", "Admission_Type"]
]

 Important Pattern 

df.loc[condition, columns]

Remember this pattern:

df.loc[
    condition,
    selected columns
]

Example:

df.loc[
    df["City"] == "Hyderabad",
    ["Patient_ID", "Surgery_Type"]
]

 Meaning:

Find patients from Hyderabad and display only Patient_ID and Surgery_Type.


 KEY POINT

loc = Label / Condition-Based Selection

 Most Important Pattern:

df.loc[condition, columns]
