```python
import pandas as pd

data = {
    "Student": ["Amit", "Neha", "Rahul", "Pooja"],
    "Marks": [85, 92, 76, 88],
    "City": ["Nagpur", "Pune", "Mumbai", "Nashik"]
}

df = pd.DataFrame(data, index=[101, 105, 110, 115])

df
```

```python
df.loc[110]  # label 101
```

```python
df.loc[101]
```

```python
df.iloc[0]
```

```python
df.iloc[0:2]
```

```python
# position 1 to 3

df.iloc[1:3]
```
